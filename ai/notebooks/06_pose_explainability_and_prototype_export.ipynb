# GymRAVANA pose explainability and prototype export

This notebook verifies the exact dataset/model-selection handoff, calculates permutation importance on the held-out collage source, and exports a local prototype package. The package is deliberately marked non-deployable.

```python
from pathlib import Path
import json
import sys

PROJECT_ROOT = next(path for path in [Path.cwd(), *Path.cwd().parents] if (path / 'artisan').exists())
sys.path.insert(0, str(PROJECT_ROOT))
from ai.pose.workflow import export_prototype

CSV_PATH = PROJECT_ROOT / 'ai/data/pose_features.csv'
METADATA_PATH = PROJECT_ROOT / 'ai/data/pose_features.metadata.json'
SELECTION_PATH = PROJECT_ROOT / 'ai/artifacts/pose_model_selection.json'
ARTIFACT_DIR = PROJECT_ROOT / 'ai/artifacts'
```

```python
artifact = export_prototype(CSV_PATH, METADATA_PATH, SELECTION_PATH, ARTIFACT_DIR)
print(json.dumps(artifact, indent=2))
assert artifact['prototype_only'] is True
assert artifact['deployment_allowed'] is False
assert artifact['task'] == 'five_class_yoga_pose_identity'
```

```python
importance_path = ARTIFACT_DIR / 'pose_identity_prototype.feature_importance.json'
importance = json.loads(importance_path.read_text(encoding='utf-8'))
print('Permutation importance on five held-out samples (descriptive only):')
for row in importance[:10]:
    print(f"{row['feature']}: {row['mean_importance']:.4f} ± {row['std_importance']:.4f}")
```

## Safe boundary

The serialized file proves that local training and export work. It must not be served to GymRAVANA members until a substantially larger, licensed, multi-person dataset passes subject/source-grouped evaluation and a trainer defines the pose taxonomy. It cannot provide medical advice or judge form correctness.
